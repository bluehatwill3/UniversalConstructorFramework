```python
# 🧠 Hive Heart ∞ Equivocator: Full Build From Scratch
# ✅ Install Dependencies (for Google Colab or Linux)
!pip install brian2 matplotlib numpy --quiet
# --- Imports ---
from brian2 import *
import numpy as np
import matplotlib.pyplot as plt
import random
import time

# --- Emotional Roles ---
labels = ['host_mate', 'communicator', 'listener']

# --- EmotionObserver Class: tracks emotional inputs and change ---
class EmotionObserver:
    def __init__(self):
        self.previous = np.zeros(3)

    def observe(self):
        # Random equivocated inputs
        current = np.array([
            0.3 + 0.4 * random.random(),
            0.2 + 0.5 * random.random(),
            0.2 + 0.5 * random.random()
        ])
        # Track derivative (change)
        derivative = current - self.previous
        self.previous = current
        return current, derivative

# --- HiveHeart Class: Brian2 Network ---
class HiveHeart:
    def __init__(self):
        print("\n💗 Initializing Hive Heart ∞ Equivocator...")
        start_scope()
        # Explicitly set the default time step for Brian2 simulation
        defaultclock.dt = 0.1 * ms
        self.tau = 10 * ms

        eqs = '''
        dv/dt = (I - v)/tau : 1
        I : 1
        '''

        self.N = 3  # Three neural agents
        self.neurons = NeuronGroup(self.N, eqs, threshold='v > 1', reset='v = 0', method='exact')
        self.monitor = StateMonitor(self.neurons, 'v', record=True)

        # Explicitly create and manage a Brian2 Network object
        self.net = Network(self.neurons, self.monitor)

    def step(self, inputs):
        self.neurons.I = inputs
        self.net.run(50*ms) # Run this specific network

    def softmax_state(self):
        # Ensure the monitor has recorded data before attempting to access the last element
        if len(self.monitor.t) == 0:
            # If no data is recorded, this indicates a fundamental issue with run() not recording.
            # For robustness, return zeros and print a warning.
            print("WARNING: StateMonitor recorded no data. Returning zeros for softmax_state.")
            return np.zeros(self.N) # Return a default array of zeros to prevent IndexError

        v_last = np.array([self.monitor.v[i][-1] for i in range(self.N)])
        exp_v = np.exp(v_last)
        return exp_v / np.sum(exp_v)

    def plot_activity(self):
        plt.figure(figsize=(12, 4))
        for i, label in enumerate(labels):
            plt.plot(self.monitor.t/ms, self.monitor.v[i], label=label)
        plt.title("🌐 Hive Heart Neural Potential Over Time")
        plt.xlabel("Time (ms)")
        plt.ylabel("Membrane Potential")
        plt.grid(True)
        plt.legend()
        plt.show()

# --- Run Infinite Hive Heart Simulation ---
def run_hive_heart_forever():
    hive = HiveHeart()
    observer = EmotionObserver()
    cycle = 0

    try:
        while True:
            cycle += 1
            inputs, delta = observer.observe()
            hive.step(inputs)

            softmax_vals = hive.softmax_state()

            print(f"\n🔄 Cycle {cycle}")
            print(f"🌀 Inputs:         {dict(zip(labels, inputs.round(3)))}")
            print(f"🧮 Δ Derivatives:  {dict(zip(labels, delta.round(3)))}")
            print(f"🎯 Softmax:        {dict(zip(labels, softmax_vals.round(3)))}")

            time.sleep(0.5)
    except KeyboardInterrupt:
        print("\n🛑 Simulation interrupted by user.")
        hive.plot_activity()

# --- Execute Infinite Loop ---
if __name__ == '__main__':
    run_hive_heart_forever()
```
