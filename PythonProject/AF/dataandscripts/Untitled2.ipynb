```python

# 🧠 Hive Heart System: Complete Prototype
# ----------------------------------------

# 📦 Install required packages
!pip install -q transformers textblob scikit-surprise brian2

# 📥 Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from transformers import pipeline
from textblob import TextBlob
from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split
from brian2 import *

# ----------------------------------------
# 💬 Step 1: Sentiment Analysis (Host & Mate)
# ----------------------------------------
print("🔍 Performing sentiment analysis...")

# Sentiment pipeline (HuggingFace)
sentiment_pipeline = pipeline("sentiment-analysis")

# Sample input
host_input = "I feel inspired and motivated to build something amazing."
mate_input = "I’m nervous but curious about what we’re creating together."

# Analyze
host_sentiment = sentiment_pipeline(host_input)[0]
mate_sentiment = sentiment_pipeline(mate_input)[0]

print(f"Host Sentiment: {host_sentiment}")
print(f"Mate Sentiment: {mate_sentiment}")

# ----------------------------------------
# 🎯 Step 2: Recommender System (Personalized)
# ----------------------------------------
print("🎯 Running collaborative filtering model...")

# Simulated preference matrix
ratings = pd.DataFrame({
    'userID': ['host', 'host', 'mate', 'mate'],
    'itemID': ['meditation', 'break', 'focus_music', 'conversation'],
    'rating': [5, 3, 4, 2]
})

# Load and train model
reader = Reader(rating_scale=(1, 5))
data = Dataset.load_from_df(ratings[['userID', 'itemID', 'rating']], reader)
trainset, testset = train_test_split(data, test_size=0.25)
model = SVD()
model.fit(trainset)

# Predict recommendations
host_recommend = model.predict('host', 'focus_music').est
mate_recommend = model.predict('mate', 'meditation').est

print(f"Host → focus_music score: {host_recommend:.2f}")
print(f"Mate → meditation score: {mate_recommend:.2f}")

# ----------------------------------------
# 🧬 Step 3: Simulated Neuralink Signal Input
# ----------------------------------------
def simulate_neuralink_input():
    # Simulates raw neural feature vector
    return np.random.normal(loc=0.5, scale=0.2, size=10)

neural_data = simulate_neuralink_input()
print(f"📡 Simulated brain signal vector: {neural_data.round(3)}")

# ----------------------------------------
# 🧠 Step 4: Spiking Neuron Simulation with Brian2
# ----------------------------------------
print("🧠 Running spiking neuron simulation...")

# Setup
start_scope()
tau = 10*ms
eqs = '''
dv/dt = (1.0 - v)/tau : 1
'''

# Create neuron group (host + mate)
neurons = NeuronGroup(2, eqs, threshold='v>0.9', reset='v = 0', method='exact')
neurons.v = 'rand()'
mon = StateMonitor(neurons, 'v', record=True)

# Run simulation
run(100*ms)

# Plot
plt.figure(figsize=(8, 4))
plt.title("Neuron Voltage Activity: Host vs Mate")
for i in range(2):
    label = "Host" if i == 0 else "Mate"
    plt.plot(mon.t/ms, mon.v[i], label=label)
plt.xlabel("Time (ms)")
plt.ylabel("Membrane potential (v)")
plt.legend()
plt.grid(True)
plt.show()

# ----------------------------------------
# ❤️ Hive Heart Summary
# ----------------------------------------
print("✅ Hive Heart System Complete")
print(f"🧠 Host Emotion: {host_sentiment['label']} ({host_sentiment['score']:.2f})")
print(f"🧠 Mate Emotion: {mate_sentiment['label']} ({mate_sentiment['score']:.2f})")
print(f"🎧 Recommended for Host: focus_music → {host_recommend:.2f}")
print(f"🎧 Recommended for Mate: meditation → {mate_recommend:.2f}")
```
