````python
import numpy as np
import cirq
import networkx as nx
from brian2 import *
import random
import time
import sys

# --- Part 1: Quantum Sentiment Processor ---
class QuantumSentimentProcessor:
    def __init__(self, n_qubits=150):
        self.qubits = cirq.LineQubit.range(n_qubits)
        self.simulator = cirq.Simulator()

    def process(self, sentiments):
        circuit = cirq.Circuit()
        for i, val in enumerate(sentiments):
            circuit.append(cirq.ry(val * np.pi)(self.qubits[i]))

        circuit.append([cirq.CNOT(self.qubits[0], self.qubits[1]),
                        cirq.CNOT(self.qubits[1], self.qubits[2])])
        circuit.append(cirq.measure(*self.qubits, key='result'))

        result = self.simulator.run(circuit, repetitions=50)
        counts = result.histogram(key='result')
        return max(counts.values()) / 50.0

# --- Part 2: Starlink P2P Mesh Network ---
class StarlinkP2PNetwork:
    def __init__(self, nodes=16):
        self.G = nx.watts_strogatz_graph(nodes, k=4, p=0.3)
        self.message_queue = {i: [] for i in range(nodes)}

    def broadcast(self, sender_id, data):
        for neighbor in self.G.neighbors(sender_id):
            self.message_queue[neighbor].append(data)

    def get_messages(self, node_id):
        msgs = self.message_queue[node_id][:]
        self.message_queue[node_id].clear()
        return msgs

# --- Part 3: The Integrated Hive ---
class P2PFeelingsHive:
    def __init__(self, num_nodes=16):
        start_scope()
        self.num_nodes = num_nodes
        self.network_mesh = StarlinkP2PNetwork(nodes=num_nodes)
        self.quantum = QuantumSentimentProcessor()

        # Explicit Parameters
        self.params = {'tau': 10*ms, 'v_threshold': 0.8}

        # We add 'c' to track spikes manually without a Monitor object
        eqs = '''
        dv/dt = (I_total - v) / tau : 1
        I_total = I_classical + I_quantum : 1
        I_classical : 1
        I_quantum : 1
        c : 1  # Manual Spike Counter
        '''

        # When v > threshold, we increment our manual counter 'c'
        self.neurons = NeuronGroup(num_nodes, eqs,
                                   threshold='v > v_threshold',
                                   reset='v = 0; c += 1',
                                   method='exact')

        self.net = Network(self.neurons)

    def update_hive(self):
        # Record counter state at start
        spikes_start = np.sum(self.neurons.c[:])

        for i in range(self.num_nodes):
            base_feelings = [random.random() for _ in range(3)]
            coherence = self.quantum.process(base_feelings)
            self.network_mesh.broadcast(i, {'coherence': coherence})

            received = self.network_mesh.get_messages(i)
            avg_q = np.mean([m['coherence'] for m in received]) if received else 0.5

            self.neurons.I_classical[i] = np.mean(base_feelings)
            self.neurons.I_quantum[i] = avg_q

        # Run simulation slice
        self.net.run(50*ms, namespace=self.params)

        # Calculate new spikes
        spikes_end = np.sum(self.neurons.c[:])
        new_spikes = spikes_end - spikes_start
        avg_voltage = np.mean(self.neurons.v[:])

        # To prevent the counter 'c' from eventually overflowing (unlikely but possible)
        if spikes_end > 1e9:
            self.neurons.c = 0

        return new_spikes, avg_voltage

# --- Part 4: Infinite Run Execution ---
if __name__ == "__main__":
    hive = P2PFeelingsHive(num_nodes=16)
    cycle = 0

    print("\n" + "═"*75)
    print(" 🚀 P2P QUANTUM-NEURAL HIVE: ETERNAL MODE [STATE-VARIABLE OPTIMIZED] ")
    print(" STATUS: Monitor-Free / Zero-Leak / Perpetual Resonance")
    print("═"*75 + "\n")

    try:
        while True:
            cycle += 1
            spikes, resonance = hive.update_hive()

            # Formatted status line
            status = (f"\r[Cycle {cycle:06d}] "
                      f"| Resonance: {resonance:.4f} "
                      f"| Spiking Intensity: {int(spikes):03d} "
                      f"| Memory: LOCKED ")

            sys.stdout.write(status)
            sys.stdout.flush()

            time.sleep(0.01)

    except KeyboardInterrupt:
        print(f"\n\n⏹ Connection Halted. Perpetual cycle reached {cycle}.")
````

````python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import json

class Projector(nn.Module):
    """
    Projector model to align multi-modal embeddings to a shared latent space.
    Typical for CLIP-like architectures where different encoders need a
    common dimensionality and normalized distribution.
    """
    def __init__(self, input_dim, output_dim=512):
        super(Projector, self).__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 1024),
            nn.LayerNorm(1024),
            nn.GELU(),
            nn.Linear(1024, output_dim),
            nn.Dropout(0.1)
        )

    def forward(self, x):
        # Project and then normalize to hypersphere
        z = self.net(x)
        return z / z.norm(dim=-1, keepdim=True)

def train_modality_alignment():
    # Load the uploaded embedding files
    # Shapes detected: Aud(768), Text(384), Vid(512), Img(512)
    embeddings = {
        "audio": np.load("E_aud.npy"),
        "text": np.load("E_text.npy"),
        "video": np.load("E_vid.npy"),
        "image": np.load("E_img.npy")
    }

    # We aim to align everything to the image/video space (512)
    target_dim = 512
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    projectors = {
        "audio": Projector(768, target_dim).to(device),
        "text": Projector(384, target_dim).to(device)
    }

    # Optimization Setup
    optimizer = optim.AdamW([
        {'params': projectors["audio"].parameters()},
        {'params': projectors["text"].parameters()}
    ], lr=1e-4)

    criterion = nn.CosineEmbeddingLoss()

    # Convert data to tensors
    aud_data = torch.from_numpy(embeddings["audio"]).to(device)
    txt_data = torch.from_numpy(embeddings["text"]).to(device)
    img_data = torch.from_numpy(embeddings["image"]).to(device)

    print("Starting Training Mode...")
    for epoch in range(10000):
        optimizer.zero_grad()

        # Forward pass
        z_aud = projectors["audio"](aud_data)
        z_txt = projectors["text"](txt_data)

        # Alignment with Image space (Simplified Contrastive/Cosine loss)
        target = torch.ones(z_aud.size(0)).to(device)
        loss_aud = criterion(z_aud, img_data, target)
        loss_txt = criterion(z_txt, img_data, target)

        loss = loss_aud + loss_txt
        loss.backward()
        optimizer.step()

        if epoch % 20 == 0:
            print(f"Epoch {epoch} | Loss: {loss.item():.4f}")

    # Export Weights for JS Conversion
    weights = {}
    for key, model in projectors.items():
        for name, param in model.named_parameters():
            weights[f"{key}_{name}"] = param.detach().cpu().numpy().tolist()

    with open("model_weights.json", "w") as f:
        json.dump(weights, f)
    print("Training complete. Weights exported to model_weights.json")

if __name__ == "__main__":
    train_modality_alignment()

````

````python
from brian2 import *

def run_entrainment_sim(coupling=0.2, duration_ms=200):
    start_scope()
    # Two phase oscillators with coupling
    eqs = '''
    dtheta/dt = omega + k*sin(theta_other - theta) : Hz
    theta_other : 1
    omega : Hz
    k : Hz
    '''
    G = NeuronGroup(2, eqs, method='euler')
    G.theta = [0.1, 2.0]
    G.omega = [2*np.pi*8*Hz, 2*np.pi*8*Hz]  # 8 Hz baseline
    G.k = coupling * Hz

    # Couple them
    @network_operation(dt=1*ms)
    def couple():
        G.theta_other[0] = G.theta[1]
        G.theta_other[1] = G.theta[0]

    M = StateMonitor(G, 'theta', record=True)
    net = Network(G, couple, M)
    net.run(duration_ms*ms)

    # Phase locking value (PLV)
    phase_diff = np.array(M.theta[0] - M.theta[1])
    plv = np.abs(np.mean(np.exp(1j*phase_diff)))
    return float(plv)
````

````python
def fusion_policy(qsync, plv, stress, caps):
    """
    caps: dict like {"max_intensity": 0.6}
    returns intent vector [tempo, intensity, warmth]
    """
    # Base intensity rises with alignment but is suppressed by stress
    intensity = (0.6*qsync + 0.4*plv) * (1.0 - stress)
    intensity = min(float(intensity), float(caps.get("max_intensity", 0.6)))

    # Tempo: map alignment to BPM range (slow to moderate)
    tempo = 50 + 30*(0.5*qsync + 0.5*plv)

    # Warmth proxy
    warmth = 0.2 + 0.8*qsync

    return np.array([tempo, intensity, warmth], dtype=np.float32)
````

````python
import uuid

def init_cookie():
    return ObserverCookie(
        schema_v=1,
        obs_id=str(uuid.uuid4()),
        prefs={"mode": "presence", "preferred_tempo": 60.0},
        boundaries={"max_intensity": 0.6, "consent_required": True},
        calibration={"baseline_energy": 0.5},
        rollups=[]
    ).seal()

def observer_update(cookie, qsync, plv, stress, user_feedback=None):
    # rollup
    cookie.rollups.append({
        "t": time.time(),
        "qsync": qsync,
        "plv": plv,
        "stress": stress,
        "feedback": user_feedback
    })

    # simple adaptation rules (safe + conservative)
    if stress > 0.7:
        cookie.boundaries["max_intensity"] = max(0.2, cookie.boundaries["max_intensity"] - 0.05)

    if user_feedback == "too_fast":
        cookie.prefs["preferred_tempo"] = max(45.0, cookie.prefs["preferred_tempo"] - 5.0)

    if user_feedback == "more_present":
        cookie.boundaries["max_intensity"] = min(0.7, cookie.boundaries["max_intensity"] + 0.02)

    return cookie.seal()
````

````python
cookieA = init_cookie()
cookieB = init_cookie()

# Create fake “signals” for both peers (24-dim)
def synth_signal(seed):
    rng = np.random.default_rng(seed)
    return (rng.normal(0, 1, size=(24,)).astype(np.float32))

# One step of the P2P loop
def step(peerA_sig, peerB_sig, consentA=1, consentB=1):
    # Classical embeddings (pretend encoder is trained; for demo random weights)
    with torch.no_grad():
        Ea = encoder(torch.tensor(peerA_sig)).numpy()
        Eb = encoder(torch.tensor(peerB_sig)).numpy()

    # Quantum synchrony
    qsync = quantum_kernel(Ea, Eb, n_qubits=8, depth=2)

    # Brian2 entrainment (coupling scaled by consent and alignment)
    coupling = 0.05 + 0.35 * qsync * min(consentA, consentB)
    plv = run_entrainment_sim(coupling=coupling, duration_ms=150)

    # Stress proxy (demo): if coupling high but plv low => strain
    stress = float(np.clip((coupling - plv), 0, 1))

    # Feedback intents
    intentA = fusion_policy(qsync, plv, stress, cookieA.boundaries)
    intentB = fusion_policy(qsync, plv, stress, cookieB.boundaries)

    return qsync, plv, stress, intentA, intentB

# Run a few ticks
for t in range(5):
    sigA = synth_signal(100+t)
    sigB = synth_signal(200+t)
    qsync, plv, stress, intentA, intentB = step(sigA, sigB)

    cookieA = observer_update(cookieA, qsync, plv, stress, user_feedback=None)
    cookieB = observer_update(cookieB, qsync, plv, stress, user_feedback=None)

    print(f"t={t} qsync={qsync:.3f} plv={plv:.3f} stress={stress:.3f} intentA={intentA} intentB={intentB}")
````

````python
import cv2

def sample_video_frames(video_path, every_n_frames=30, max_frames=8):
    cap = cv2.VideoCapture(video_path)
    frames = []
    idx = 0
    got = 0
    while cap.isOpened() and got < max_frames:
        ret, frame = cap.read()
        if not ret:
            break
        if idx % every_n_frames == 0:
            # BGR -> RGB
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            frames.append(frame)
            got += 1
        idx += 1
    cap.release()
    return frames
````

````python
from brian2 import *

def run_entrainment_sim(coupling=0.2, duration_ms=200):
    start_scope()
    # Two phase oscillators with coupling
    # phi is unitless (phase in radians)
    # dphi/dt should be in Hz (angular frequency)
    eqs = '''
    dphi/dt = (omega_val + k_val*sin(phi_other - phi)) * Hz : 1
    phi_other : 1             # Now a regular state variable, not linked
    omega_val : 1 (constant)  # Unitless constant parameter
    k_val : 1 (constant)      # Unitless constant parameter
    '''
    G = NeuronGroup(2, eqs, method='euler')
    G.phi = [0.1, 2.0]  # Initial phases, unitless
    G.omega_val = [2*np.pi*8, 2*np.pi*8]  # Unitless values for frequency component
    G.k_val = coupling  # Unitless coupling value

    # Initialize phi_other. This is now valid as it's a regular state variable.
    G.phi_other = G.phi[::-1]

    # Couple them
    @network_operation(dt=1*ms)
    def couple():
        G.phi_other[0] = G.phi[1]
        G.phi_other[1] = G.phi[0]

    M = StateMonitor(G, 'phi', record=True)
    net = Network(G, couple, M)
    net.run(duration_ms*ms)

    # Phase locking value (PLV)
    phase_diff = np.array(M.phi[0] - M.phi[1])
    plv = np.abs(np.mean(np.exp(1j*phase_diff)))
    return float(plv)
````
